```python
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from IPython.display import display
# Step 1: Initialize a Quantum Circuit with 1 qubit
# Qubits default to the North Pole (|0⟩ state)
qc = QuantumCircuit(1)
# Step 2: Apply a Hadamard (H) gate to create a superposition
# This rotates the vector 90 degrees around the Y-axis to land on the X-axis

qc.h(0)
# Step 3: Extract the state vector from the circuit
state = Statevector.from_instruction(qc)
# Step 4: Display the numeric state vector coefficients
print("Quantum State Vector:")
print(state)
print("\nRendering Bloch Sphere...")
# Step 5: Force Jupyter to display the 3D visualization inline
display(state.draw('bloch'))
```
